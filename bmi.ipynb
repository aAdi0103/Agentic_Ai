```python
from langgraph.graph import StateGraph, START, END
# START and END are special nodes that represent the start and end of the workflow. It defines the entry and end point of the graph.
from typing import TypedDict
```

```python
# This class represents a State in the workflow. It defines the structure of the state that will be passed between nodes in the graph.
# State is the data that is passed between nodes in the graph. Each nodes can read and write to the state and returns the updated state to the next node in the graph.

class BMI(TypedDict):

    weight_kg: float
    height_m: float
    bmi: float
    category: str
```

```python
def bmi_calculator(state:BMI) -> BMI:
    weight = state["weight_kg"]
    height = state["height_m"]
    bmi = weight / (height ** 2)
    state["bmi"] = bmi
    return state
```

```python
def category_calculator(state:BMI) -> BMI:
    bmi = state["bmi"]
    if bmi < 18.5:
        category = "Underweight"
    elif 18.5 <= bmi < 24.9:
        category = "Normal weight"
    elif 25 <= bmi < 29.9:
        category = "Overweight"
    else:
        category = "Obesity"
    state["category"] = category
    return state
```

```python
graph = StateGraph(BMI)

graph.add_node("bmi_cal", bmi_calculator)
graph.add_node("category_cal", category_calculator)

graph.add_edge(START, "bmi_cal")
graph.add_edge("bmi_cal", "category_cal")
graph.add_edge("category_cal", END)

# compile means to convert the graph into a workflow that can be executed. It checks for any errors in the graph and prepares it for execution.
workflow = graph.compile()
```

```python
initial_state = {'weight_kg': 70, 'height_m': 1.75}
output = workflow.invoke(initial_state)
print(output)
```
