```python
from langgraph.graph import StateGraph,START, END
from typing import TypedDict
```

```python
class attributes(TypedDict):
    runs: int
    ball: int
    four: int
    six: int
    sr: float
    boundary_percentage: float
    ballperboundary: float
    summary:str
```

```python
def sr(state: attributes)-> attributes:
    strr = state['runs']/state['ball']
    return {'sr': strr}
```

```python
def boundary_percentage(state: attributes)-> attributes:
    bdry_percentage = (((state['four']*4) + (state['six']*6))/state['runs'])*100
    return {'boundary_percentage': bdry_percentage}
```

```python
def ballperboundary(state: attributes)-> attributes:
    ballperbdry = (state['four'] + state['six'])/state['ball']
    return {'ballperboundary': ballperbdry}
```

```python
def summary(state:attributes)-> attributes:
    smmry= f"""
   sr: {state['sr']}
   boundary_percentage: {state['boundary_percentage']}
   ballperboundary: {state['ballperboundary']}
"""
    return {'summary':smmry}
```

```python
graph = StateGraph(attributes)
graph.add_node('sr',sr)
graph.add_node('boundary_percentage',boundary_percentage)
graph.add_node('ballperboundary',ballperboundary)
graph.add_node('summary', summary)

graph.add_edge(START,'boundary_percentage')
graph.add_edge(START,'sr')
graph.add_edge(START,'ballperboundary')

graph.add_edge('boundary_percentage','summary')
graph.add_edge('ballperboundary','summary')
graph.add_edge('sr','summary') 
graph.add_edge('summary',END) 

workflow = graph.compile()
```

```python
initial_state = {'runs':100,'ball':60,'four':6,'six':4}
final_state = workflow.invoke(initial_state)
print(final_state)
```
